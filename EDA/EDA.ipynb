```python
df[['Individual_Fare','Fare']].describe()
```

```python
df['Family_Size'] = df['SibSp'] + df['Parch'] + 1
```

```python
df['Family_Size']
```

```python
df
```

```python
def family_describe(num):
  if num == 1:
    return 'alone'
  elif num > 1 and num <= 4:
    return 'small '
  else:
    return 'large'
```

```python
df['Family_Type'] = df['Family_Size'].apply(family_describe)
```
