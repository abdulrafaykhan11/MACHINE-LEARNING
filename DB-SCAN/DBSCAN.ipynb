```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import make_moons
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans,DBSCAN
```

```python
X,y_true = make_moons(n_samples=500,noise=0.05,random_state=42)
```

```python
df =pd.DataFrame(X,columns=['feature1','feature2'])
```

```python
from numpy._core.multiarray import scalar
scaler = StandardScaler()
X_scaled = scaler.fit_transform(df)
```

```python
kmeans = KMeans(n_clusters=2,random_state=42)
kmeans_label = kmeans.fit_predict(X_scaled)
```

```python
df['kmeans_cluster'] = kmeans_label
```

```python
sns.scatterplot(data=df,x='feature1',y='feature2',hue='kmeans_cluster',palette='tab10')
```

```python
dbscan = DBSCAN(eps=0.3,min_samples=5)
dbscan_labels = dbscan.fit_predict(X_scaled)
```

```python
df['dbscan_clusters'] = dbscan_labels
```

```python
sns.scatterplot(data=df,x='feature1',y='feature2',hue=dbscan_labels,palette='tab10')
```
