```python
import numpy as np
from scipy.stats import norm
```

```python
sample = [172,174,168,169,171,173,175,170,169,172]
population_mean = 170
population_std = 3
x = np.mean(sample)
n = len(sample)
```

```python
z_score = (x-population_mean)/(population_std/np.sqrt(n))
```

```python
z_score
```

```python
p_value = 2 * (1-norm.cdf(abs(z_score)))
```

```python
p_value
```

```python
alpha = 0.05

if p_value < alpha:
  print("I will reject the null hypothesis")
else:
  print("I will accept the null hypothesis")
```
