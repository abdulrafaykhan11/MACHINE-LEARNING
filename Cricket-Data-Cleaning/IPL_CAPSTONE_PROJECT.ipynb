```python
total_Win = df['match_winner'].value_counts()
sns.barplot(y=total_Win.index,x=total_Win.values,palette='rainbow')
plt.title('Total Win')

```

```python
sns.countplot(x = df['toss_decision'],palette='rainbow',hue=df['match_winner'])
plt.title("Toss Decision Trends ")
```

```python
count = df[df['match_winner'] == df['toss_winner']]['match_id'].count()
percentage = (count * 100)/df.shape[0]
percentage.round(2)
```

```python
sns.countplot(x=df['won_by'])
```

```python
man_of_match = df['player_of_the_match'].value_counts().head(10)
sns.barplot(y=man_of_match.index,x=man_of_match.values,palette='rainbow')
plt.title('Man of the Match')
```

```python
high = df.groupby('top_scorer')['highscore'].sum().sort_values(ascending=False).head(2)
high
high.plot(kind='barh',color='red')
plt.title('High Score')
```

```python
df.head()
```

```python
df['highest_wicket'] = df['best_bowling_figure'].apply(lambda x : x.split('--')[0])
df['highest_wicket'] = df['highest_wicket'].astype(int)
bowler = df.groupby('best_bowling')['highest_wicket'].sum().sort_values(ascending=False).head(10)
bowler.plot(kind='barh',color='green')
plt.title('Highest Wicket')
```

```python
df['venue'].value_counts().head(10).plot(kind='barh',color='red')
plt.title('Venue')
```
