```python
import numpy as np
from scipy.stats import chi2_contingency
import seaborn as sns
import pandas as pd
```

```python
df = sns.load_dataset("titanic")
```

```python
df
```

```python
contingency_table = pd.crosstab(df['sex'],df['survived'])
```

```python
contingency_table
```

```python
chi2 , p_value , dof , expected = chi2_contingency(contingency_table)
```

```python
alpha = 0.05
```

```python
if p_value < alpha :
  print("Reject the null hypothesis and there is a significant relation between sex and survival")
else:
  print("there is no connection")
```
