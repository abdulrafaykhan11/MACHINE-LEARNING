```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = sns.load_dataset('iris')
```

```python
df
```

```python
df['species'].unique()
```

```python
from sklearn.model_selection import train_test_split
```

```python
X = df.drop('species', axis=1)
y = df['species']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.33, random_state=42)
```

```python
from sklearn.neighbors import KNeighborsClassifier
```

```python
model_knn = KNeighborsClassifier(n_neighbors=5)
```

```python
model_knn.fit(X_train, y_train)
```

```python
y_pred_knn = model_knn.predict(X_test)
```

```python
model_knn.score(X_test, y_test)
```

```python
from sklearn.svm import SVC
```

```python
model_svm = SVC(C=30,gamma='auto',kernel='rbf')
```

```python
model_svm.fit(X_train, y_train)
```

```python
model_svm.score(X_test, y_test)
```

```python
from sklearn.model_selection import GridSearchCV
```

```python
classifier = GridSearchCV((model_svm),{
    'C' : [1,10,20],
    'kernel' : ['rbf','linear']
}, cv=5, return_train_score=False
)
```

```python
 classifier.fit(X,y)
```

```python
classifier.cv_results_
```

```python
results = pd.DataFrame(classifier.cv_results_)
```

```python
results
```

```python
results[['param_C','param_kernel','mean_test_score']]
```

```python
classifier2 = GridSearchCV((model_knn) , {
    'n_neighbors' : [1,5,10],
    'algorithm' : ['auto', 'ball_tree', 'kd_tree', 'brute'],
    'metric' : ['minkowski', 'euclidean', 'manhattan'],
    'weights' : ['uniform', 'distance'],
}, cv=5, return_train_score=False)
```

```python
classifier2.fit(X,y)
```

```python
classifier2.cv_results_
```

```python
result_knn = pd.DataFrame(classifier2.cv_results_)
```

```python
result_knn
```

```python
result_knn[['param_algorithm','param_metric','param_n_neighbors','param_weights','mean_test_score']]
```
